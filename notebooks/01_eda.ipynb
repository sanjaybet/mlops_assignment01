# Heart Disease Dataset: Acquisition and EDA

This notebook uses the repository's acquisition and EDA modules so notebook results stay consistent with the automated pipeline.

The original target values are converted to a binary label: `0` means no disease, while `1` through `4` mean disease present. Missing feature values are reported here and imputed later inside the cross-validation model pipeline.

```python
from pathlib import Path
import os
import sys

CURRENT_DIR = Path.cwd().resolve()
PROJECT_ROOT = next(
    folder for folder in (CURRENT_DIR, *CURRENT_DIR.parents)
    if (folder / "src" / "eda.py").exists()
)
os.chdir(PROJECT_ROOT)
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from src.download_data import RAW_DATA_PATH, download_dataset
from src.eda import load_raw_data, run_eda

print(f"Project root: {PROJECT_ROOT}")
```

```python
if not (PROJECT_ROOT / RAW_DATA_PATH).exists():
    download_dataset(output_path=str(PROJECT_ROOT / RAW_DATA_PATH))

raw_df = load_raw_data(str(PROJECT_ROOT / RAW_DATA_PATH))
cleaned_df = run_eda()

print(f"Raw shape: {raw_df.shape}")
print(f"Cleaned shape: {cleaned_df.shape}")
print("Missing values before model-pipeline imputation:")
display(raw_df.isna().sum().rename("missing_count").to_frame())
print("Binary target counts:")
display(cleaned_df["target"].value_counts().sort_index().rename("records").to_frame())
```
