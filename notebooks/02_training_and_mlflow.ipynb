## Compare recorded runs

The table below reads the latest runs from the local MLflow tracking store. Each run includes CV/holdout metrics, parameters, tuning output, plots, and a model artifact.

```python
import mlflow

runs = mlflow.search_runs(
    experiment_names=['Heart_Disease_Prediction'],
    order_by=['start_time DESC'],
)
metric_columns = [
    column for column in runs.columns
    if column.startswith(('metrics.cv_mean_', 'metrics.test_'))
]
parameter_columns = [
    column for column in runs.columns
    if column in ('tags.mlflow.runName', 'params.selection_metric')
]
display(runs[parameter_columns + metric_columns].head(4))
```
